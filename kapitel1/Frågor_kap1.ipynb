E. Grid Search är en metod för att hitta en bra kombination av hyperparametrar genom att testa olika kombinationer systematiskt. Grid Search är en metod för att hitta bra hyperparametrar till en modell. Man anger olika värden för hyperparametrarna och Grid Search testar alla möjliga kombinationer av dessa. Namnet kommer från att kombinationerna bildar ett "grid" (rutnät) och att man sedan söker igenom detta rutnät efter den bästa kombinationen.

Till exempel kan man testa olika värden för max_depth och n_estimators. Grid Search testar då alla kombinationer och använder exempelvis k-delad korsvalidering för att utvärdera dem.

refit=True betyder att när Grid Search har hittat de bästa hyperparametrarna så tränas modellen om på hela datasetet med dessa parametrar. Den färdiga modellen kan sedan användas för att göra nya prediktioner.


F. Kategorisk data är data som består av olika kategorier eller grupper istället för numeriska värden.

Det finns två typer av kategorisk data: nominal data och ordinal data.

Nominal data är kategorier som inte har någon naturlig ordning, till exempel färger: röd, blå och grön. För nominal data kan man använda one-hot-encoding eller dummy-variable-encoding, där kategorierna omvandlas till binära värden, alltså 0 och 1.

Ordinal data är kategorier som har en naturlig ordning, till exempel liten, medium och stor. För detta kan man använda ordinal encoding, där kategorierna omvandlas till numeriska värden, exempelvis liten = 1, medium = 2 och stor = 3.

Skillnaden är alltså att nominal data saknar ordning medan ordinal data har en ordning.

G. Feature engineering är processen av att skapa, förändra eller välja ut variabler från Raw Data för att göra en modell bättre på att jitta mönster. Det kan till exempel vara att skapa en ny feature genom att kombinera befintliga features, omvandla kategorisk data eller ta bort features som inte är relevanta.

9.  MSE ät egentligen bättre ju lägre det är, medan scikit-learn vill att ett högre scoring-värde alltid ska betyda bättre model
MSE mäter hur stora felen är mellan modellens förutsägelser och de faktiska värdena. För MSE gäller att ett lägre värde är bättre.
I scikit-learn är principen däremot att ett högre scoring-värde ska innebära bättre prestanda. Därför används neg_mean_squared_error, där MSE-värdet multipliceras med -1.
Exempelvis blir MSE = 10 till -10 och MSE = 25 till -25. Eftersom -10 är högre än -25 kommer scikit-learn att bedöma modellen med MSE 10 som bättre.
Det negativa värdet används alltså för att vända på skalan så att "higher is better"-principen fungerar.
