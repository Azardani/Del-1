1. Vid klassificering ska modellen förutsäga vilken kategori en observation tillhör.

Exempel:

Spam eller inte spam
Sjuk eller frisk
Bild på katt eller hund
Kreditansökan: godkänd eller nekad
Kund kommer att lämna företaget eller stanna

Det skiljer sig från regression eftersom regression förutsäger ett numeriskt kontinuerligt värde, medan klassificering förutsäger en klass.

2. OvR = en modell per klass mot alla andra. OvO = en modell för varje par av klasser.

3. 
A. En confusion matrix visar hur modellens förutsägelser fördelar sig mellan:

TP – True Positive
TN – True Negative
FP – False Positive
FN – False Negative

Den gör det lätt att se vilka typer av fel modellen gör.

B. Accuracy är andelen korrekta förutsägelser
C. Precision svarar på: Av alla som modellen förutspådde som positiva, hur många var faktiskt positiva?
D. Recall svarar på: Av alla faktiskt positiva, hur många hittade modellen?
E. F1-score kombinerar precision och recall genom deras harmoniska medel. F1 är användbart när man vill ha en balans mellan precision och recall, särskilt när klasserna är obalanserade.
F. ROC-kurvan visar sambandet mellan:
True Positive Rate (TPR) = recall
False Positive Rate (FPR)
när man ändrar klassificeringströskeln.
En modell med bra förmåga att skilja klasserna åt får en ROC-kurva som ligger närmare det övre vänstra hörnet. ROC-AUC används ofta som ett sammanfattande måt

5. 
A. Logistisk regression beräknar sannolikheten att en observation tillhör en viss klass och använder sedan en tröskel för att bestämma klassen.
Trots namnet används den alltså för klassificering, inte vanlig regression
B. SVM försöker hitta en beslutsgräns som separerar klasserna och samtidigt skapar så stor marginal som möjligt mellan klasserna.
C. Ett beslutsträd delar upp datan steg för steg genom frågor om features.
D. Ensemble learning innebär att man kombinerar flera modeller för att skapa en samlad prediktion.
E. Random Forest är en ensemble av många beslutsträd
Varje träd tränas på en bootstrap-sample och använder dessutom slumpmässiga urval av features vid splits
F. Extra Trees liknar Random Forest eftersom även den bygger på många beslutsträd.
Skillnaden är att Extra Trees introducerar mer slumpmässighet när splits väljs

6. Med trädmodeller kan vi undersöka vilka features som modellen använder mest för att göra sina beslut.
Det såg vi faktiskt i diamantuppgiften där carat fick klart högst importance

7. Precision och recall måste vägas mot varandra. I rättsväsendet är falska positiva särskilt allvarliga eftersom en oskyldig person kan pekas ut och dömas. Därför bör man inte bara maximera precision eller recall utan ta hänsyn till konsekvenserna av både falska positiva och falska negativa.

9. På träningsdatan använder vi .fit_transform() eftersom preprocessorn först behöver lära sig parametrarna från träningsdatan och sedan transformera den. På validerings- och testdatan använder vi endast .transform() eftersom vi ska använda samma parametrar som lärdes från träningsdatan. Om vi använder .fit_transform() där riskerar vi data leakage, eftersom preprocessorn då lär sig från data som ska användas för utvärdering.
