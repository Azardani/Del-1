1. Curse of dimensionality innebär att när antalet dimensioner/features ökar blir datan glesare och ML-modeller får svårare att hitta mönster.
 Det kan leda till högre beräkningskostnad och ökad risk för overfitting

2. Dimensionsreducering innebär att man minskar antalet dimensioner i ett dataset samtidigt som man försöker behålla den viktigaste informationen. Det görs för att förenkla datan, minska beräkningskostnaden och ibland förbättra modellens prestanda

3. PCA reducerar antalet dimensioner genom att skapa nya dimensioner, principal components, 
som är riktningar där datan varierar mest. Den första komponenten förklarar mest variation, den andra näst mest och så vidare.
 Genom att behålla de viktigaste komponenterna kan man minska antalet dimensioner samtidigt som man behåller mycket av informationen

5. bäst prediktion är inte automatiskt samma sak som bäst modell. En modell måste också vara tillräckligt snabb och praktiskt användbar. 
Det stämmer inte att man alltid bara vill maximera prediktionsprestandan. Tid är också en viktig aspekt, både vid träning och vid prediktion. En något sämre modell kan därför vara mer lämplig om den är betydligt snabbare och billigare att använda

6. Efter PCA ersätts de ursprungliga variablerna av nya principal components som är kombinationer av de ursprungliga variablerna. Det gör datan mindre lättolkad eftersom det blir svårare att se vilken ursprunglig variabel som påverkar resultatet. PCA ger alltså dimensionsreduktion men kan försämra modellens tolkbarhet.

8. Koden skapar först en dataset med 1000 observationer och 3 features. Med PCA reduceras datan från 3 till 2 dimensioner genom att behålla de två viktigaste principal components. Därefter används inverse_transform() för att försöka återskapa datan till 3 dimensioner. Eftersom en dimension togs bort har information gått förlorad, och den återskapade datan är därför inte exakt samma som originaldatan.
