1. Regressionsproblem innebär att man försöker förutsäga ett kontinuerligt numeriskt värde. Exempel är att förutsäga huspriser, löner, temperatur eller försäljning.

2. MSE beräknar medelvärdet av de kvadrerade felen och straffar därför stora fel kraftigt. 
RMSE är roten ur MSE och har samma enhet som det värde som förutsägs. 
MAE beräknar medelvärdet av de absoluta felen och är mindre känsligt för stora avvikelser.

3. Nej. RMSE är roten ur MSE och eftersom roten är en monoton funktion behåller modellerna samma rangordning. Skillnaden är att RMSE är lättare att tolka eftersom det har samma enhet som målvariabeln.

4. Gradient descent är en optimeringsalgoritm som stegvis ändrar modellens parametrar i riktningen som minskar kostnadsfunktionen, tills man förhoppningsvis når ett minimum.

5. Bias-variance trade-off handlar om balansen mellan modellens bias och variance. En för enkel modell kan ha hög bias och underfitta datan, medan en för komplex modell kan ha hög variance och överfitta datan. Målet är därför att hitta en modell som är tillräckligt komplex för att fånga sambanden i datan, men ändå tillräckligt simpel att den lär sig brus och detaljer från träningsdatan.

6. 
A. Linjär regression modellerar sambandet mellan variabler genom en linjär funktion och försöker hitta koefficienter som ger så små prediktionsfel som möjligt.
B. Ridge regression är linjär regression med L2-regularisering. Den straffar stora koefficienter och kan därför minska risken för overfitting.
C. Lasso regression är linjär regression med L1-regularisering. Den kan göra vissa koefficienter till noll och därmed automatiskt välja bort vissa variabler.
D. Elastic Net kombinerar L1- och L2-regularisering från Lasso och Ridge. Det gör att modellen både kan minska stora koefficienter och sätta vissa koefficienter till noll.
E. SVR försöker hitta en funktion som passar datan inom en viss felmarginal. Med hjälp av kernels kan SVR även hantera icke-linjära samband.
F. Ett beslutsträd delar upp datan stegvis baserat på olika villkor. För regression ger lövet en numerisk prediktion, ofta baserad på värdena hos träningsobservationerna som hamnat där.
G. Ensemble learning innebär att flera modeller kombineras för att skapa en gemensam prediktion. VotingRegressor kombinerar prediktionerna från flera olika regressionsmodeller, medan BaggingRegressor tränar flera modeller på olika urval av träningsdatan.
H. Random Forest är en ensemble av många beslutsträd. Träden tränas med slumpmässiga urval av data och features, och deras prediktioner kombineras för att få den slutliga prediktionen. Detta kan ge en mer robust modell och minska risken för overfitting jämfört med ett enskilt beslutsträd.

8. Bagging använder slumpmässiga urval av träningsdatan med återläggning, medan pasting använder slumpmässiga urval utan återläggning. I båda fallen tränas flera modeller på olika delar av datan och deras resultat kombineras

11. R², eller determinationskoefficienten, är ett mått på hur stor andel av variationen i den beroende variabeln som modellen kan förklara. Ett R²-värde på 1 innebär perfekt förklaring. R² = 0 innebär att modellen inte förklarar mer variation än en modell som alltid förutsäger medelvärdet. R² kan även vara negativt om modellen presterar sämre än denna baslinjemodel
