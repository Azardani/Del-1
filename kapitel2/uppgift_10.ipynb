```python
import pandas as pd

df = pd.read_csv("salary_dataset.csv")

print(df.head())
print(df.columns)
```

```python
X = df[["YearsExperience"]]
y = df["Salary"]

print(X.head())
print(y.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import cross_validate
```

```python
linear_model = LinearRegression()
tree_model = DecisionTreeRegressor(random_state=42)
```

```python
linear_cv = cross_validate(
    linear_model,
    X_train,
    y_train,
    scoring="neg_root_mean_squared_error",
    cv=5
)

tree_cv = cross_validate(
    tree_model,
    X_train,
    y_train,
    scoring="neg_root_mean_squared_error",
    cv=5
)
```

```python
linear_rmse = -linear_cv["test_score"].mean()
tree_rmse = -tree_cv["test_score"].mean()

print("Linear Regression CV RMSE:", linear_rmse)
print("Decision Tree CV RMSE:", tree_rmse)
```

```python
import numpy as np
from sklearn.metrics import mean_squared_error

best_model = LinearRegression()

best_model.fit(X_train, y_train)

test_predictions = best_model.predict(X_test)

test_rmse = np.sqrt(mean_squared_error(y_test, test_predictions))

print("Linear Regression Test RMSE:", test_rmse)
```
