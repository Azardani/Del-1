#1. De sju stegen i ett ML-projekt är:#

1. Definiera problemet och skapa en helhetsbild.
2. Få tillgång till datan.
3. Utforska datan genom exploratory data analysis (EDA).
4. Bearbeta datan.
5. Genomföra ML-modellering.
6. Presentera lösningen för intressenter.
7. Produktionssätta modellen och övervaka implementeringen.

I verkligheten följs stegen inte som en rak progression. Arbetet är generellt mer iterativt, vilket innebär att man kan behöva gå fram och tillbaka mellan olika steg. Exempelvis kan man upptäcka när man undersöker datan att den data som behövs saknas, och då kan man behöva gå tillbaka och ändra målet med projektet. Checklistan är därför ett stöd i arbetet och ska inte följas slaviskt.

#2. Att produktionssätta en modell betyder att man tar den färdiga ML-modellen från utvecklingsmiljön och gör den tillgänglig för användning i en riktig applikation eller verksamhet

#3 Scikit-learn är ett Python-bibliotek för maskininlärning. Biblioteket följer fyra centrala designprinciper: konsistens, inspektion, begränsning av klasser samt rimliga standardvärden för hyperparametrar.
En estimator är det centrala objektet i scikit-learn och lär sig från data genom metoden .fit().
En predictor är en estimator som även kan göra prediktioner genom .predict() och har en .score()-metod för att mäta hur bra prediktionerna är.
En transformer är en estimator som kan transformera data genom .transform(). Den kan även använda .fit_transform(), vilket utför både fit och transform i ett steg.

Keras är ett högnivå-API som gör det enklare att skapa och träna neurala nätverk. Keras används tillsammans med TensorFlow och ger ett mer lättanvänt gränssnitt.

#6 Många AI/ML-projekt misslyckas eftersom man från början kan ha otydliga eller orealistiska mål, dålig eller otillräcklig data eller för höga förväntningar på vad AI kan lösa. Det kan också visa sig att en modell fungerar bra tekniskt men inte skapar någon verklig nytta för verksamheten.
Vi bör därför förhålla oss till AI/ML-projekt som iterativa processer. Man behöver testa sina antaganden tidigt, arbeta med prototyper och vara beredd att ändra både mål och lösning när man lär sig mer om problemet och datan. Ett misslyckat försök kan därför också ge värdefull information som hjälper projektet framåt.
