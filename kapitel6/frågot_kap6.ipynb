1. Klustring innebär att man grupperar datapunkter utifrån hur lika de är varandra. Det används exempelvis för kundsegmentering, bildanalys, marknadsföring och att hitta avvikelser

2. K-means börjar med ett bestämt antal kluster och placerar en centroid i varje kluster. Varje datapunkt kopplas till den närmaste centroiden. Därefter räknas medelvärdet av punkterna i varje kluster ut och centroiden flyttas dit. Detta upprepas tills klustren stabiliseras.

3. För att välja antal kluster kan man testa flera olika värden på K. Med inertia undersöker man hur kompakt klustren är och använder ofta elbow-metoden för att hitta en punkt där ytterligare kluster ger liten förbättring. Silhouette score visar hur väl datapunkterna passar sina kluster och hur väl klustren är separerade. Ett högre silhouette-värde är bättre. Ett silhouette diagram kan användas för att visuellt jämföra olika K och se hur bra datapunkterna är grupperade. Genom att kombinera inertia, elbow-diagram och silhouette score/diagram kan man välja ett lämpligt antal kluster.
